# number of rows = n
# number of stars = row number

```python
for i in range(1,n+1):
    for j in range(1,i+1):
        print("*", end="")
    print()
    
for i in range(n,0,-1):
    for j in range(1,i+1):
        print("*", end="")
    print()
```

```python
k = int(input("Enter Any number :"))
```

```python
is_prime = True

for i in range(2,k):
    if k%i==0:
        is_prime = False
        break
        
if is_prime:
    print("The given number {} is prime".format(k))
else:
    print("The given number {} is not prime".format(k))
```
