```python
import string
import random
```

```python
DB = {}
```

```python
def getShortURL(longURL):
    """
    Given : LongURL
    Result : ShortURL
    """
    chars = string.ascii_lowercase
    lgth = random.randint(4,6)
    shortURL = "".join([random.choice(chars) for i in range(lgth)])
    
    if shortURL in DB:
        return getShortURL(longURL)
    else:
        DB[shortURL] = longURL
        
    result = "cm.in/" + shortURL
    return result
```

```python
# Reverse - Get Long URL
```

```python
def getLongURL(shortURL):
    """
    Given : shortURL
    result : LongURL
    """
    
    key = shortURL.split('/')[-1]
    
    
    if key in DB:
        return DB[key] 
    else:
        return "ShortURL doesn't exists"
    
```

```python
def getShortURL(longURL, myShortURL = None):
    """
    Given : longURL
    result : ShortURL
    """
    if myShortURL:
        if myShortURL in DB:
            return "Provided ShortURL already taken.."
        else:
            DB[myShortURL] = longURL
            result = "cm.in/" + myShortURL
            return result
    
    chars = string.ascii_lowercase + string.digits
    l = random.randint(4,6)
    shortURL = "".join([random.choice(chars) for i in range(l)])
    
    if shortURL in DB:
        return getShortURL(longURL)
    else:
        DB[shortURL] = longURL
    
    result = "cm.in/" + shortURL
    
    return result
    
    
    
    
```

```python
# updating long url
```

```python
def updateLongURL(shortURL, newLongURL):
    """
    it will update the newLong URL 
    """
    
    if shortURL in DB:
        shortURL = shortURL.split("/")[-1]
        DB[shortURL] = newLongURL
        result = "cm.in/" + shortURL
        return result
    else:
        return "Provided ShortURL doesn't exist"
    
```
