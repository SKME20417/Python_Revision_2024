```python
p_word = input("Enter any word :")
```

```python
# Method 1


rev_word = p_word[::-1]

if (p_word == rev_word):
    print(f"The given word {p_word} is plaindrome")
else:
    print(f"The given word {p_word} is not palindrome")
```

```python
p_word = input("Enter any word :")
```

```python
is_palindrome = True
start = 0
end = len(p_word)-1


while (start<=end):
    if p_word[start]!=p_word[end]:
        is_palindrome = False
        break
        
    start+=1
    end-=1
    
if (is_palindrome):
    print(f"The given word {p_word} is plaindrome")
else:
    print(f"The given word {p_word} is not palindrome")
```
