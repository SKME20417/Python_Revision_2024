```python
class Human:
    #class variables
    
    database = []
    id_seq = 101
    population = 0
    
    def __init__(self, name, age, is_alive = True):
        print("New object is created...")
        self.name = name
        self.age = age
        self.is_alive = is_alive
        
        self.id = Human.id_seq
        Human.id_seq+=1
        Human.population+=1
        Human.database.append(self)
    def introduce(self):
        print(f"Hi my name is {self.name} and i am {self.age} years old")
        
    def die(self):
        if self.is_alive:
            print(self.name,"I am dying...")
            self.is_alive = False
            Human.population-=1
        else:
            print("The person is already dead..")
        
    def __repr__(self):
        """
        This is represent an object in readable format
        """
        return f"[{self.id}, {self.name}, {self.age}, {self.is_alive}]"
```

```python
class Hitman(Human):
    def __init__(self, name, age):
        super().__init__(name, age)
        self.kills = 0
        self.kill_list = []
        
    def killing(self, person):
        if person.is_alive:
            print(self.name, "is killing...", person.name)
            person.die()
            self.kills+=1
            self.kill_list.append(person.name)
        else:
            print("The person is alread dead/ killed by someone", person.name)
        
    def introduce(self):
        print(f"Hi my name is {self.name} and i have killed {self.kills} people")
```
