```python
# reading from a file
```

```python
with open("extension.txt", mode='r') as f:
    data = f.read()
    print(data)
```

```python
with open("extension.txt", mode='r') as f :
    d1 = f.read(5)
    print(d1)
    
    d2 = f.read(4)
    print(d2)
    
```

```python
# The file handle
```

```python
with open('extension.txt', mode='r') as f:
    print(f.tell())
    
    print(f.read(4))
    
    print(f.tell())
    
    print(f.read(6))
    
    print(f.tell())
    
    print(f.seek(5))
    
    print(f.tell())
    
    print(f.read(16))
    
    
```
