```python
# Multilevel Inheritance
```

```python
class A:
    def m(self):
        print("I am inside class A..")

class B(A):
    def m1(self):
        print("I am inside class B...")
        
class C(B):
    def m1(self):
        print("C class has over ridding of m1")
            
    def m2(self):
        print("I am inside class C")
        
class D(C):
    def m3(self):
        print("I am inside class d")
        
    def m2(self):
        print("m2 has been over ridden...")
```

```python
a = A()
b = B()
c = C()
d = D()
```

```python
a.m()
```

```python
b.m()
```

```python
b.m1()
```

```python
c.m()
```

```python
c.m1()
```

```python
c.m2()
```

```python
d.m()
```

```python
d.m1()
```

```python
d.m2()
```

```python
d.m3()
```

```python
# Multiple Inheritance
```

```python
class A(object):
    def m(self):
        print("I am inside class A..")

class B:
    def m1(self):
        print("I am inside class B...")
        
class C:
    def m1(self):
        print("C class has over ridding of m1")
            
    def m2(self):
        print("I am inside class C")
        
class D(C,A,B):
    def m3(self):
        print("I am inside class d")
        
    def m2(self):
        print("m2 has been over ridden...")
```

```python
a = A()
b = B()
c = C()
d = D()
```

```python
a.m()
```

```python
b.m1()
```

```python
c.m1()
```

```python
c.m2()
```

```python
d.m()
```

```python
d.m1()
```

```python
d.m2()
```

```python
d.m3()
```

```python
# class/object level Polymorphism
```

```python
class Animal:
    def speak(self):
        print("I am speaking something....")
        
class Cat(Animal):
    def speak(self):
        print("I am doing meaouwww...")
        
class Dog(Animal):
    def speak(self):
        print("I will bark....")
        
class Duck(Animal):
    def speak(self):
        print("I will quack....")
```

```python
animal = Animal()
cat = Cat()
dog = Dog()
dick = Duck()
```

```python
animal.speak()
```

```python
cat.speak()
```

```python
for i in [animal, cat, dog, dick]:
    i.speak()
```
