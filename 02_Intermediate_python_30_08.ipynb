```python
# Custom Iterator:
```

```python
class CubeNumber:
    def __init__(self, number = 0):
        self.number = number
    
    def __iter__(self):
        self.ele = 0
        return self
    
    def __next__(self):
        if self.ele<=self.number:
            result = self.ele**3
            self.ele+=1
            return result
        else:
            raise StopIteration
```

```python
done = CubeNumber()
```

```python
done
```

```python
done.number
```

```python
my_iter = done.__iter__()
```

```python
my_iter.__next__()
```

```python
def my_generator():
    
    n = 1
    print("First Time...")
    yield n
    
    n+=1
    print("Second time..")
    yield n
    
    
    n+=1
    print("Third Time...")
    yield n
    
    n+=1
    print("Forth Time..")
    yield n
    
    
```

```python
# Return Vs Yield
```

```python
def nml_fun(n):
    i = 1
    while i<=n:
        return i
        i+=1
    
```

```python
def spl_gen(n):
    i = 1
    while (i<=n):
        yield i
        i+=1
```

```python
# reversing a string
```

```python
def reverse_string(text):
    for i in range(len(text)-1, -1, -1):
        yield text[i]
```

```python
for i in reverse_string("machine"):
    print(i)
```

```python
# decorators
```

```python
def basic():
    print("I am a very basic function")
```

```python
basic()
```

```python
def makespecial(func):
    def wrapper():
        print("*"*50)
        print("New features are added")
        func()
        print("="*50)
    return wrapper
    
```

```python
basic()
```

```python
new_func = makespecial(basic)
```

```python
new_func()
```

```python
@makespecial
def simple():
    print("i am very simple")
```

```python
simple()
```

```python
# decorating functions with parametrs
```

```python
DB = {
    'ravi':"ravi@123",
    'mohit':"mohit@123",
    'prateek':'prateek@123'
}
```

```python
def add(a,b):
    c = a+b
    print(c)
```

```python
add(5,6)
```

```python
add(9,10)
```

```python
def login_required(func):
    def wrapper(username, password, *args, **kwargs):
        if username in DB and DB[username] == password:
            print("Authenticated successfully..")
            func(*args, **kwargs)
        else:
            print("failed : Try again")
    return wrapper
```

```python
@login_required
def add(a,b):
    c = a+b
    print(c)
```

```python
add(5,3)
```

```python
add(10,2)
```

```python
add('ravi', "ravi@123", 8,9)
```

```python
# Chaining Decorators in Python
```

```python
def doller(func):
    def wrapper(*args, **kwargs):
        print("$"*50)
        func(*args, **kwargs)
        print("$"*50)
    return wrapper
```

```python
def star(func):
    def wrapper(*args, **kwargs):
        print("*"*50)
        func(*args, **kwargs)
        print("*"*50)
    return wrapper
```

```python
def cap(func):
    def wrapper(*args, **kwargs):
        print("^"*50)
        func(*args, **kwargs)
        print("^"*50)
    return wrapper
```

```python
@cap
@doller
@star
def main(text):
    print(text)
```

```python
main("python is very funy..")
```
