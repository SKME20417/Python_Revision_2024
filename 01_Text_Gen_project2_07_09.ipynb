```python
def generateTable(data, k = 4):
    """
    Transition Table : This will generate transition Table
    """
    
    T = {}
    
    for i in range(len(data)-k):
        X = data[i:i+k]
        Y = data[i+k]
        
        if T.get(X) is None:
            T[X] = {}
            T[X][Y] = 1
        else:
            if T[X].get(Y) is None:
                T[X][Y] = 1
            else:
                T[X][Y]+=1
    return T
```

```python
T = generateTable("hello hello helli helly")
```

```python
def convertFreqIntoProb(T):
    for kx in T.keys():
        s = sum(T[kx].values())
        for k in T[kx].keys():
            T[kx][k] = T[kx][k]/s
    return T
```

```python
def read_text(filepath):
    with open(filepath, encoding='cp437') as f:
        my_data = f.read().lower()
        return my_data
```

```python
# Train Our Markov Chains
```

```python
def trainMarkovChain(text, k = 4):
    T = generateTable(text, k)
    T = convertFreqIntoProb(T)
    
    return T
```

```python
model = trainMarkovChain(text)
```

```python
def sample_next(context, T, k):
    context = context[-k:]
    
    if T.get(context) is None:
        return ""
    
    possible_chars = list(T.get(context).keys())
    possible_probabs = list(T.get(context).values())
    return random.choices(possible_chars, weights=possible_probabs)[0]
```

```python
sample_next("hello water between ", model, 4)
```

```python
def generateText(starting_sent, T, k = 4, max_len = 100):
    sentence = starting_sent
    context = starting_sent[-k:]
    
    for i in range(max_len):
        next_pred = sample_next(context, T, k)
        sentence+=next_pred
        context = sentence[-k:]
    return sentence
```
