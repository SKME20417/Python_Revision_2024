```python
def product(lst):
    """
    parametrs : list
    result : it will accept list and multiply all the numbers
    """
    res = 1
    for i in lst:
        res*=i
    return res
```

```python
def LCM(a,b):
    """
    This function will return LCM
    """
    max_num = a if a>b else b
    
    while True:
        if (max_num%a==0) and (max_num%b==0):
            return max_num
        else:
            max_num+=1
```

```python
LCM(6,7)
```

```python
LCM(3,5)
```

```python
LCM(4,16)
```

```python
# Arbitary Arguments(args)
```

```python
def args_func(*args):
    print(type(args))
    print(args)
    
    for i in args:
        print(i)
```

```python
args_func(14,25,36,True, False, 5+7j, [7,8,9], (10,12,13))
```

```python
def introduction(**kwargs):
    print(kwargs)
    print(type(kwargs))
    
    for k,v in kwargs.items():
        print(k, ":", v)
```

```python
introduction(name = "tanzeel", age = 30, nationality = 'indian', company = "tgc")
```

```python
introduction(name = 'muskaan', age = 40, company = "google", bard = "hp", hobbies = ['coding', 'dancing', 'singing']) 
```

```python
def demo(a,b,c,age = 18, *args, **kwargs):
    print(a,b,c,age)
    print(args)
    print(kwargs)
```
