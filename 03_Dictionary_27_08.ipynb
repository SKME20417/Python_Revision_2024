```python
# Dictionary Compreshension
```

```python
dic = {'a':1, 'b':2, 'c':3, 'd':4, 'e':5}
```

```python
dic
```

```python
dic
```

```python
for k,v in dic.items():
    print(k, ":", v)
```

```python
{k:v for k,v in dic.items()}
```

```python
{k:v for k,v in dic.items() if v>=2}
```

```python
s = {}
for k,v in dic.items():
    if v>=2:
        s[k] = v
```

```python
s
```

```python
dic
```

```python
{k+'z' : v**3 for k,v in dic.items()}
```

```python
dic
```

```python
# reversing a dictionary
```

```python
dic
```

```python
rev = {}
for k,v in dic.items():
    rev[v] = k
```

```python
rev
```
