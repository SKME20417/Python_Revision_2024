```python
steps = int(input("Enter Number of steps :"))
```

```python
t = [1,2,3,4,5,6]
```

```python
t
```

```python

l = len(t)-1

for _ in range(steps):
    key = t[-1]
    
    for j in range(l,0,-1):
        t[j] = t[j-1]
    
    t[0] = key
    
```
